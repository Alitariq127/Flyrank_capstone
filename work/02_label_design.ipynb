# 02 — Label Design (notes)

## Unit of analysis
**Page-level**, not page x query. Query-level would be more granular but far sparser
and noisier at this pass; page-level keeps the action ("rewrite this page") directly
actionable for a content team.

## Primary metric
**Clicks** (not impressions or CTR alone) — most decision-relevant: it's what a
content team is ultimately trying to move, and it already nets out ranking x CTR.

## Windows
- **Feature window:** weeks 0-11 (12 weeks) — everything the model is allowed to see.
- **Label window:** weeks 12-15 (4 weeks) — used ONLY to compute the label,
  split further into an early half and late half to measure change.
- **Test window:** weeks 16-19 (4 weeks, held out) — untouched until final evaluation.

## Label logic (see `_lib.compute_labels`)
Compare early-half vs late-half clicks within the label window:
- `growing`: late > early by >=15%
- `declining`: late < early by >=15%
- `stable`: within +/-15%
- `recovering`: was `declining` in the *previous* label period, `growing` now
  (derived, not primary — see `_lib.mark_recovering`)

The 15% threshold is deliberately not >0% — a stricter bar filters out
week-to-week noise from being mislabeled as a real trend.

## Leakage discipline
Feature window and label window must never overlap, and no feature is allowed
to peek into the label or test windows. This is asserted explicitly in
`03_label_pipeline.ipynb` before anything else runs.
